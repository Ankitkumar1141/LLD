```python
## Animal: Parent Class
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def eat(self):
        print(f"{self.name} eats food")
```

```python
## Dog: Child Class (Animal)
class Dog(Animal):
    def __init__(self, name,age,breed):
        super().__init__(name,age)
        self.breed = breed

    def bark(self):
        print(f"{self.name} of breed {self.breed} barks loudly! ")
```

```python
dog1=Dog("Tommy",2,"Labrador")
print(dog1.name)
print(dog1.age)
print(dog1.breed)
```

```python
### Parent class: Employee
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def display_info(self):
        print(f"{self.name} has {self.salary} INR.")
```

```python
## Child Class- Manager
class Manager(Employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.dept = department

    def manage_team(self):
        print(f"{self.name} manages {self.dept} team very professionally and experiencly.")
```

```python
emp1 = Manager("Rohan",100000,"Growth")
print(emp1.name)
print(emp1.dept)
emp1.display_info()
emp1.manage_team()
```

```python
### Vehicle: Parent Class
class Vehicle:
    def __init__(self,brand,speed):
        self.brand = brand
        self.speed = speed

## Car: Child class
class Car(Vehicle):
    def __init__(self,brand,speed,model,fuel_type):
        super().__init__(brand,speed)
        self.model = model
        self.fuel_type = fuel_type

    def display_car(self):
        print(f"{self.brand}")
        print(f"{self.speed}")
        print(f"{self.model}")
        print(f"{self.fuel_type}")
```

```python
car1 = Car("Toyota", 180, "Fortuner", "Diesel")

car1.display_car()
```

```python
### Bank Account: Parent Class
class Account:
    def __init__(self, account_number,balance):
        self.__account_number = account_number
        self.__balance = balance

    ### getter for account number
    def get_account_number(self):
        return self.__account_number

    ## getter for balance
    def get_balance(self):
        return self.__balance

    def deposit_money(self, amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print(f"{amount} is sucessfully deposited")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")
        else:
            print("Invalid withdrawal or insufficient balance")
```

```python
# Child Class
class SavingsAccount(Account):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.__interest_rate = interest_rate

    # Getter for interest rate
    def get_interest_rate(self):
        return self.__interest_rate

    # Calculate interest
    def calculate_interest(self):
        interest = self.get_balance() * self.__interest_rate / 100
        return interest
```

```python
account = SavingsAccount("12345", 10000, 5)

account.deposit_money(2000)
account.withdraw(3000)

print("Account Number:", account.get_account_number())
print("Balance:", account.get_balance())
print("Interest Rate:", account.get_interest_rate())
print("Interest:", account.calculate_interest())
```
