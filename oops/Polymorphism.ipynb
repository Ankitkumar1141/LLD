```python
class Area:
    def calculate_area(self):
        return 0


class Rectangle(Area):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        area = self.length * self.width
        return f"Area: {area}"


class Circle(Area):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        area = 3.14 * self.radius ** 2
        return f"Area: {area}"


class Triangle(Area):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def calculate_area(self):
        area = 0.5 * self.base * self.height
        return f"Area: {area}"
```

```python
rectangle = Rectangle(10, 5)
circle = Circle(7)
triangle = Triangle(8, 6)
```

```python
shapes = [rectangle, circle, triangle]
```

```python
for shape in shapes:
    print(shape.calculate_area())
```

```python
class Employee:
    def calculate_salary(self):
        return 0

class FullTimeEmployee(Employee):
    def __init__(self, name, monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary
    def calculate_salary(self):
        return f"Salary: {self.monthly_salary}"

class PartTimeEmployee(Employee):
    def __init__(self,name,hours, hourly_rate):
        self.name = name
        self.hours = hours
        self.hourly_rate = hourly_rate
    def calculate_salary(self):
        return f"Salary: {self.hours * self.hourly_rate}"

class ContractEmployee(Employee):
    def __init__(self, name, project_payment):
        self.name = name
        self.salary = project_payment
    def calculate_salary(self):
        return f"Salary: {self.salary}"
```

```python
employees = [
    FullTimeEmployee("Rahul", 50000),
    PartTimeEmployee("Amit", 100, 500),
    ContractEmployee("Priya", 75000)
]

for employee in employees:
    print(employee.calculate_salary())
```

```python
class Payment:
    def process_payment(self):
        return 0
class CreditCard(Payment):
    def __init__(self, card_holder, card_number):
        self.holder = card_holder
        self.number = card_number

    def process_payment(self):
        return f"Payment process started using Credit Card"

class Upi(Payment):
    def __init__(self,upi_id):
        self.upi = upi_id

    def process_payment(self):
        return f"Payment process started using UPI"

class Netbanking(Payment):
    def __init__(self, account_number,bank_name):
        self.number = account_number
        self.bank = bank_name
    def process_payment(self):
        return f"Payment process has started using Net Banking"
```

```python
payments = [
    CreditCard("Rahul", "123456789"),
    Upi("rahul@upi"),
    Netbanking("987654321", "SBI")
]

for payment in payments:
    print(payment.process_payment())
```
