##### Bank Account Security

```python
class BankSecurity:
    def __init__(self, account_holder, account_number, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.__balance = balance  # Private attribute

    def deposit(self,amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print(f"{amount} deposited sucessfully to your account")
        else:
            print("Invalid amount.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance = self.__balance - amount
            print(f"{amount} withdrawn successfully from your account")
        else:
            print("Invalid amount or insufficient funds.")

    def get_balance(self):
        print(f"Your current balance is: {self.__balance}")

    def display_account(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Account Number: {self.account_number}")
        print(f"Balance: {self.__balance}")
```

```python
account = BankSecurity("Ankit", "1234567890", 1000)
```

```python
account.deposit(1000)
```

```python
account.withdraw(500)
```

```python
account.get_balance()
```

##### Employee Salary Protection

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.id = emp_id
        self.__salary = salary  # Private attribute

    def get_salary(self):
        return self.__salary
    
    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
            print(f"Salary updated to: {self.__salary}")
        else:
            print("Invalid salary amount.")

    def increase_salary(self, increment):
        if increment > 0:
            self.__salary = self.__salary + (self.__salary * (increment/100))
            print(f"Salary increased by {increment}%. New salary: {self.__salary}")
        else:
            print("Invalid increment percentage.")
    
    def display_employee(self):
        print(f"Employee Name: {self.name}")
        print(f"Employee ID: {self.id}")
        print(f"Salary: {self.__salary}")
```

```python
emp = Employee("Ankit", 101, 50000)

print(emp.get_salary())

emp.set_salary(60000)
emp.increase_salary(10)

emp.display_employee()
```

##### Digital wallet System

```python
class Wallet:
    def __init__(self, owner,wallet_id, balance, pin):
        self.owner = owner
        self.wallet_id = wallet_id
        self.__balance = balance
        self.__pin = pin

    def deposit(self, amount):
        if amount >0:
            self.__balance += amount
            print(f"{amount} deposited successfully.")
        else:
            print("Invalid amount.")
    
    def check_balance(self):
        pin = input("Enter your PIN to check balance: ")
        if pin == self.__pin:
            print(f"Your current balance is: {self.__balance}")
        else:
            print("Incorrect PIN. Access denied.")
    
    def withdraw(self, amount):
        pin = input("Enter your PIN to withdraw: ")
        if pin == self.__pin:
            if amount > 0 and amount <= self.__balance:
                self.__balance -= amount
                print(f"{amount} withdrawn successfully.")
            else:
                print("Invalid amount or insufficient funds.")
        else:
            print("Incorrect PIN. Access denied.")

    def change_pin(self):
        current_pin = input("Enter your current PIN: ")
        if current_pin == self.__pin:
            new_pin = input("Enter your new PIN: ")
            confirm_pin = input("Confirm your new PIN: ")
            if new_pin == confirm_pin:
                self.__pin = new_pin
                print("PIN changed successfully.")
            else:
                print("New PIN and confirmation do not match.")
        else:
            print("Incorrect current PIN. Access denied.")

    def display_wallet(self):
        print(f"Wallet Owner: {self.owner}")
        print(f"Wallet ID: {self.wallet_id}")
        print(f"Balance: {self.__balance}")
```

```python
wallet = Wallet("Ankit", "W101", 10000, "1234")
```

```python
wallet.deposit(5000)
```

```python
wallet.check_balance()
```

```python
wallet.check_balance()
```

```python
wallet.withdraw(2000)
```

```python
wallet.change_pin()
```

```python
wallet.check_balance()
```

##### E-Commerce Product Price Prediction

```python
class Product:
    def __init__(self,name, product_id, price, stock):
        self.name = name
        self.product_id = product_id
        self.__price = price
        self.__stock = stock

    def get_price(self):
        return self.__price

    def set_price(self):
        new_price = float(input("Enter the new price: "))
        if new_price > 0:
            self.__price = new_price
            print(f"Price updated to: {self.__price}")
        else:
            print("Invalid price amount.")

    def add_stock(self):
        additional_stock = int(input("Enter the quantity to add to stock: "))
        if additional_stock > 0:
            self.__stock += additional_stock
            print(f"Stock increased by {additional_stock}.")
        else:
            print("Invalid stock quantity.")

    def remove_stock(self):
        remove_quantity = int(input("Enter the quantity to remove from stock: "))
        if 0 < remove_quantity <= self.__stock:
            self.__stock -= remove_quantity
            print(f"Stock decreased by {remove_quantity}.")
        else:
            print("Invalid stock quantity or insufficient stock.")

    def get_stock(self):
        return self.__stock
    
    def apply_discount(self, discount_percentage):
        if 0 < discount_percentage < 100:
            self.__price = self.__price - (self.__price * (discount_percentage / 100))
        else:
            print("Invalid discount percentage.")

    def display_product(self):
        print(f"Product Name: {self.name}")
        print(f"Product ID: {self.product_id}")
        print(f"Price: {self.__price}")
        print(f"Stock: {self.__stock}")
```

```python
product = Product("Laptop", "P101", 100000, 20)
```

```python
product.apply_discount(20)
product.remove_stock()
product.add_stock()
product.set_price()
product.display_product()
```
