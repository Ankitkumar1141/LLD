##### 1. Student Class

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display_data(self):
        print(f"Name: {self.name}, Age: {self.age}, Marks: {self.marks}")
```

```python
### object of student
student1= Student("Ankit",20,98)
student1.display_data()
```

##### 2. Product

```python
class Product:
    def __init__(self, name, price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return  self.price*self.quantity

    def apply_discount(self, discount_percentage):
        if 0 <= discount_percentage <= 100:
             self.price -= (self.price * discount_percentage) / 100
        else:
            print("Invalid discount percentage")

    def display_data(self):
        print(f"Product Name: {self.name}, Price: {self.price}, Quantity: {self.quantity}, Total Stock Value: {self.total_value()}")
```

```python
### object of product class
product1 = Product("Laptop", 1000, 5)
product1.apply_discount(-7)# Applying a 10% discount
product1.display_data()
```

##### 3. Employee 

```python
class Employee:
    def __init__(self, name, id, base_salary,experience, performance_score):
        self.name = name
        self.id = id
        self.base_salary = base_salary
        self.experience = experience
        self.performance_score = performance_score

    def calculate_bonus(self):
        if self.experience >5:
            return self.base_salary*0.15
        elif self.experience >2 and self.experience <5:
            return self.base_salary*0.10
        else:
            return self.base_salary*0.05

    def get_rating(self):
        if self.performance_score >=0 and self.performance_score <= 100:
            if self.performance_score >= 90 and self.performance_score <= 100:
                return "Excellent"
            elif self.performance_score >= 75 and self.performance_score < 90:
                return "Good"
            elif self.performance_score >= 60 and self.performance_score < 75:
                return "Average"
            else:
                return "Poor"
        else:
            return "Invalid performance score"

    def update_performance_score(self, new_score):
        if 0 <= new_score <= 100:
            self.performance_score = new_score
        else:
            print("Invalid performance score")

    def calculate_salary(self):
        return self.base_salary + self.calculate_bonus()

    def promote(self):
        self.base_salary = self.base_salary*0.20 + self.base_salary
        self.experience = self.experience + 1

    def display_employee_data(self):
        print(f"Employee Name: {self.name}")
        print(f"Employee ID: {self.id}")
        print(f"Base Salary: {self.base_salary}")
        print(f"Experience: {self.experience}")
        print(f"Performance Score: {self.performance_score}")
        print(f"Rating: {self.get_rating()}")
        print(f"Bonus: {self.calculate_bonus()}")
        print(f"Total Salary: {self.calculate_salary()}")
```

```python
ankit = Employee("Ankit","EMP001",20000,5,89)
```

```python
ankit.calculate_bonus()
ankit.promote()
```

```python
ankit.display_employee_data()
```

```python
ankit.update_performance_score(91)
```

##### Library Management System

```python
class Book:
    def __init__(self, title, author, isbn, ):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True

    def borrow_book(self):
        if self.is_available == False:
            print(f"The book '{self.title}' is currently not available for borrowing.")
        else:
            self.is_available = False
            print(f"The book '{self.title}' has been borrowed.")

    def return_book(self):
        if self.is_available == True:
            print(f"The book '{self.title}' is already available in the library.")
        else:
            self.is_available = True
            print(f"The book '{self.title}' has been returned.")

    def display_book_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"ISBN: {self.isbn}")
        print(f"Availability: {self.is_available}")
```
