##### 2. Product

```python
class Product:
    def __init__(self, name, price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return  self.price*self.quantity

    def apply_discount(self, discount_percentage):
        if 0 <= discount_percentage <= 100:
             self.price -= (self.price * discount_percentage) / 100
        else:
            print("Invalid discount percentage")

    def display_data(self):
        print(f"Product Name: {self.name}, Price: {self.price}, Quantity: {self.quantity}, Total Stock Value: {self.total_value()}")
```

```python
### object of product class
product1 = Product("Laptop", 1000, 5)
product1.apply_discount(-7)# Applying a 10% discount
product1.display_data()
```

##### 3. Employee 

```python
class Employee:
    def __init__(self, name, id, base_salary,experience, performance_score):
        self.name = name
        self.id = id
        self.base_salary = base_salary
        self.experience = experience
        self.performance_score = performance_score

    def calculate_bonus(self):
        if self.experience >5:
            return self.base_salary*0.15
        elif self.experience >2 and self.experience <5:
            return self.base_salary*0.10
        else:
            return self.base_salary*0.05

    def get_rating(self):
        if self.performance_score >=0 and self.performance_score <= 100:
            if self.performance_score >= 90 and self.performance_score <= 100:
                return "Excellent"
            elif self.performance_score >= 75 and self.performance_score < 90:
                return "Good"
            elif self.performance_score >= 60 and self.performance_score < 75:
                return "Average"
            else:
                return "Poor"
        else:
            return "Invalid performance score"

    def update_performance_score(self, new_score):
        if 0 <= new_score <= 100:
            self.performance_score = new_score
        else:
            print("Invalid performance score")

    def calculate_salary(self):
        return self.base_salary + self.calculate_bonus()

    def promote(self):
        self.base_salary = self.base_salary*0.20 + self.base_salary
        self.experience = self.experience + 1

    def display_employee_data(self):
        print(f"Employee Name: {self.name}")
        print(f"Employee ID: {self.id}")
        print(f"Base Salary: {self.base_salary}")
        print(f"Experience: {self.experience}")
        print(f"Performance Score: {self.performance_score}")
        print(f"Rating: {self.get_rating()}")
        print(f"Bonus: {self.calculate_bonus()}")
        print(f"Total Salary: {self.calculate_salary()}")
```

```python
ankit = Employee("Ankit","EMP001",20000,5,89)
```

```python
ankit.calculate_bonus()
ankit.promote()
```

```python
ankit.display_employee_data()
```

```python
ankit.update_performance_score(91)
```

##### Library Management System

```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True

    def borrow_book(self):
        if self.is_available == False:
            print(f"The book '{self.title}' is currently not available for borrowing.")
        else:
            self.is_available = False
            print(f"The book '{self.title}' has been borrowed.")

    def return_book(self):
        if self.is_available == True:
            print(f"The book '{self.title}' is already available in the library.")
        else:
            self.is_available = True
            print(f"The book '{self.title}' has been returned.")

    def display_book_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"ISBN: {self.isbn}")
        print(f"Availability: {self.is_available}")
```

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self,book):
        self.books.append(book)
        print(f"The book '{book.title}' has been added to the library.")

    def find_books(self, title):
        for book in self.books:
            if book.title == title:
                return book
        return None

    def search_book(self, title):
        book = self.find_books(title)
        if book:
            book.display_book_info()
        else:
            print(f"The book '{title}' is not available in the library.")

    def borrow_books(self, title):
        book = self.find_books(title)
        if book:
            book.borrow_book()
        else:
            print(f"The book '{title}' is not available in the library.")

    def return_books(self, title):
        book = self.find_books(title)
        if book:
            book.return_book()
        else:
            print(f"The book '{title}' is not available in the library.")

    def display_all_books(self):
        for book in self.books:
            book.display_book_info()

```

```python
book1 = Book("Clean Code", "Robert C. Martin", "9780132350884")
book2 = Book("Python Crash Course", "Eric Matthes", "9781593279288")
book3 = Book("The Hobbit", "J.R.R. Tolkien", "9780345339683")
```

```python
library = Library()
```

```python
### Adding books to the library
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
```

##### Banking system

```python
class BankAccount:
    def __init__(self, account_number, holder_name, balance):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self,amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}")
        else:
            print("Invalid withdrawal amount or insufficient balance")

    def transfer_amt(self,amount, account):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            account.balance += amount
            print(f"Transferred {amount} to {account.holder_name}")
        else:
            print("Invalid transfer amount or insufficient balance")

    def display_info(self):
        print(f"Account Number: {self.account_number}")
        print(f"Holder Name: {self.holder_name}")
        print(f"Balance: {self.balance}")
```

```python
class Bank:
    def __init__(self):
        self.accounts = []

    def add_account(self, account):
        self.accounts.append(account)
        print(f"Account for {account.holder_name} added to the bank.")

    def find_account(self, account_number):
        for account in self.accounts:
            if account.account_number == account_number:
                return account
        return None

    def transfer(self, from_account, to_account, amount):
        from_acc = self.find_account(from_account)
        to_acc = self.find_account(to_account)
        if from_acc and to_acc:
            from_acc.transfer_amt(amount, to_acc)
        else:
            print("One or both accounts not found.")

    def display_all_accounts(self):
        for account in self.accounts:
            account.display_info()
             
```

```python
account1 = BankAccount("ACC101", "Ankit", 10000)
account2 = BankAccount("ACC102", "Rahul", 5000)

bank = Bank()

bank.add_account(account1)
bank.add_account(account2)

bank.transfer("ACC101", "ACC102", 3000)

bank.display_all_accounts()
```
