#### Hospital Management System 

-- Using Encapsulation and Inheritance

                    Patient
                   /       \
                  /         \
        InPatient       OutPatient

```python
class Patient:
    def __init__(self, name, age, patient_id, phone,blood_group):
        self.__name=name
        self.__age = age
        self.__patient_id = patient_id
        self.__phone= phone
        self.__blood_group = blood_group
    
    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age
    def get_patient_id(self):
        return self.__patient_id
    def get_phone(self):
        return self.__phone
    def get_blood_group(self):
        return self.__blood_group

    ## setter
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid age. Age must be greater than 0.")

    def set_phone(self, phone):
        if len(str(phone)) == 10:
            self.__phone = phone
        else:
            print("Invalid phone number. It must contain 10 digits.")

    # Display patient information
    def display_patient(self):
        print("Name:", self.__name)
        print("Age:", self.__age)
        print("Patient ID:", self.__patient_id)
        print("Phone:", self.__phone)
        print("Blood Group:", self.__blood_group)
```

```python
# Child Class - InPatient
class InPatient(Patient):
    def __init__(
        self,
        name,
        age,
        patient_id,
        phone,
        blood_group,
        room_number,
        days_admitted,
        doctor_name,
        room_charge
    ):
        super().__init__(name, age, patient_id, phone, blood_group)

        self.__room_number = room_number
        self.__days_admitted = days_admitted
        self.__doctor_name = doctor_name
        self.__room_charge = room_charge

    # Getters
    def get_room_number(self):
        return self.__room_number

    def get_days_admitted(self):
        return self.__days_admitted

    def get_doctor_name(self):
        return self.__doctor_name

    def get_room_charge(self):
        return self.__room_charge

    # Setter
    def set_days_admitted(self, days):
        if days >= 0:
            self.__days_admitted = days
        else:
            print("Days admitted cannot be negative.")

    # Calculate room bill
    def calculate_room_bill(self):
        return self.__days_admitted * self.__room_charge

    # Display inpatient information
    def display_inpatient(self):
        self.display_patient()
        print("Room Number:", self.__room_number)
        print("Days Admitted:", self.__days_admitted)
        print("Doctor Name:", self.__doctor_name)
        print("Room Charge:", self.__room_charge)
        print("Total Room Bill:", self.calculate_room_bill())
```

```python
class OutPatient(Patient):
    def __init__(
        self,
        name,
        age,
        patient_id,
        phone,
        blood_group,
        doctor_name,
        appointment_date,
        consultation_fee
    ):
        super().__init__(name, age, patient_id, phone, blood_group)

        self.__doctor_name = doctor_name
        self.__appointment_date = appointment_date
        self.__consultation_fee = consultation_fee

    # Getters
    def get_doctor_name(self):
        return self.__doctor_name

    def get_appointment_date(self):
        return self.__appointment_date

    def get_consultation_fee(self):
        return self.__consultation_fee

    # Display outpatient information
    def display_outpatient(self):
        self.display_patient()
        print("Doctor Name:", self.__doctor_name)
        print("Appointment Date:", self.__appointment_date)
        print("Consultation Fee:", self.__consultation_fee)
```

```python

inpatient = InPatient(
    "Rahul",
    45,
    "P101",
    "9876543210",
    "B+",
    205,
    5,
    "Dr. Sharma",
    2500
)

print("===== INPATIENT DETAILS =====")
inpatient.display_inpatient()

print("\n===== UPDATING DAYS =====")
inpatient.set_days_admitted(7)

print("Updated Days:", inpatient.get_days_admitted())
print("Updated Room Bill:", inpatient.calculate_room_bill())

print("\n===== INVALID AGE TEST =====")
inpatient.set_age(-5)
```

```python
print("\n===== OUTPATIENT DETAILS =====")

outpatient = OutPatient(
    "Priya",
    30,
    "P102",
    "9123456780",
    "O+",
    "Dr. Verma",
    "20-09-2026",
    800
)

outpatient.display_outpatient()
```
